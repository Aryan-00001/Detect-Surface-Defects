```python
from ultralytics import YOLO

# Load a model architecture (to train from scratch)
# Using 'yolov8n.yaml' means you are loading the architecture, not pre-trained weights.
model = YOLO("yolov8n.yaml")

# Train the model
results = model.train(
    data="/content/drive/MyDrive/DEFECTED SURFACE DETECTION /data.yaml",  # Path to your data config
    epochs=50,
    imgsz=640,
    batch=16,
    name="neu_det_v8_from_scratch",
)
```

```python
from ultralytics import YOLO

# Load your trained model
model = YOLO("/content/drive/MyDrive/DEFECTED SURFACE DETECTION /runs/detect/neu_det_v8_from_scratch/weights/best.pt")

# Path to your test image
image_path = "/content/drive/MyDrive/DEFECTED SURFACE DETECTION /NEU-DET/test/images/scratches_274.jpg"

# Run inference
results = model.predict(source=image_path, conf=0.25, save=True)
# conf=0.25 sets the confidence threshold; save=True saves the output image

# Display results (optional, in notebook)
results[0].show()

# Print detected objects with bounding boxes
for box in results[0].boxes:
    x1, y1, x2, y2 = box.xyxy[0]  # Bounding box coordinates
    conf = box.conf[0]            # Confidence score
    cls = int(box.cls[0])          # Class index
    print(f"Class ID: {cls}, Confidence: {conf:.2f}, Box: [{x1:.0f}, {y1:.0f}, {x2:.0f}, {y2:.0f}]")
```
